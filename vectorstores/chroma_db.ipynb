```python
import warnings
warnings.filterwarnings('ignore', category=DeprecationWarning)

from llm_llamacpp import embedder
from langchain_chroma import Chroma
from langchain_core.documents import Document
```

```python
doc1 = Document(
    page_content='Counter Strike 2 is fun to watch',
    metadata={'genre':'FPS'}
)

doc2 = Document(
    page_content="If you have a computer and you haven't played Skyrim, you just haven't lived life",
    metadata={'genre':'RPG'}
)

doc3 = Document(
    page_content='Rise of Nations is a really fun game. Strategy lovers should try it out',
    metadata={'genre':'RTS'}
)

docs = [doc1, doc2, doc3]
```

```python
chroma_vectorstore = Chroma(
    embedding_function=embedder,
    persist_directory='chroma_db',
    collection_name='Games_sample'
)
```

```python
chroma_vectorstore.add_documents(docs)
```

```python
chroma_vectorstore.get(include=['documents', 'embeddings', 'metadatas'])
```

```python
chroma_vectorstore.similarity_search(
    query='RPG Games',
    k=1
)
```

```python
chroma_vectorstore.similarity_search(
    query='RTS',
    k=1
)
```

```python
chroma_vectorstore.similarity_search(
    query='',
    filter={'genre':'RPG'}
)
```
