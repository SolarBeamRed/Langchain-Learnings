```python
import warnings
warnings.filterwarnings('ignore', category=DeprecationWarning)

from llm_llamacpp import embedder
from langchain_community.vectorstores import FAISS
from langchain_core.documents import Document
```

```python
doc1 = Document(
    page_content='Counter Strike 2 is fun to watch',
    metadata={'genre':'FPS'}
)

doc2 = Document(
    page_content="If you have a computer and you haven't played Skyrim, you just haven't lived life",
    metadata={'genre':'RPG'}
)

doc3 = Document(
    page_content='Rise of Nations is a really fun game. Strategy lovers should try it out',
    metadata={'genre':'RTS'}
)

docs = [doc1, doc2, doc3]
```

```python
faiss_vectorstore = FAISS.from_documents(
    documents=docs,
    embedding=embedder
)

faiss_vectorstore.save_local('faiss_db')
```

```python
faiss_vectorstore.similarity_search(
    query='rpg'
)
```

```python
faiss_vectorstore.similarity_search(
    query='',
    filter={'genre':'RTS'},
)
```
