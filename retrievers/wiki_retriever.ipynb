```python
import warnings
warnings.filterwarnings('ignore', category=DeprecationWarning)

from langchain_community.retrievers import WikipediaRetriever
from llm_llamacpp import llm
```

```python
wiki_retriever = WikipediaRetriever(
    top_k_results=2,
    lang='en'
)
```

```python
docs = wiki_retriever.invoke('The Elder Scrolls video games')
```

```python
print(len(docs))
```

```python
print(docs[0].metadata)
```

```python
print(docs[1].metadata)
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

template = PromptTemplate(
    template='Tell me what this document is about in a summary of about 50-75 words:\n{document}',
    input_variables=['document'],
    validate_template=True
)

str_parser = StrOutputParser()
```

```python
summaries = []
chain = template | llm | str_parser

for doc in docs:
    result = chain.invoke({'document': doc.page_content})
    summaries.append(result)
```

```python
print(len(summaries))
```

```python
for i, summary in enumerate(summaries):
    print(f'Summary for doc {i+1}:\n{summary}', end='\n\n\n')
```
