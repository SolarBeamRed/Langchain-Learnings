```python
from langchain_chroma import Chroma
from langchain_core.documents import Document

from llm_llamacpp import embedder
```

```python
docs = [
    Document(page_content='TES V Skyrim takes place in Skyrim, a beautiful, snow filled region. It is mostly inhabited by the Nords'),
    Document(page_content='TES IV Oblivion takes place in Cyrodiil, the province mostly inhabited by Imperials'),
    Document(page_content='TES III Morrowing takes place in Morrowind, and is primarily inhabited by the racist, miserable Dark Elves')
]
```

```python
vectorstore = Chroma.from_documents(
    documents=docs,
    embedding=embedder,
    collection_name='TES_Vector_collection'
)
```

```python
retriever = vectorstore.as_retriever(search_kwargs={'k':2})
```

```python
results = retriever.invoke('cold')
```

```python
print(len(results))
```

```python
print('Sorted according to relevance, the retrieved documents are:')
for i, doc in enumerate(results):
    print(f'Document no {i+1}\n{doc}', end='\n\n')
```

```python
results = retriever.invoke('Instruct: Given a query, retrieve relevant passages that answer the query.\nQuery: racism')
```

```python
print('Sorted according to relevance, the retrieved documents are:')
for i, doc in enumerate(results):
    print(f'Document no {i+1}\n{doc}', end='\n\n')
```
